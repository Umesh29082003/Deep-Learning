```python
def initialize_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    l = len(layer_dims)            # number of layers in the network
    for i in range(1, l):
        parameters['W' + str(i)] = np.ones((layer_dims[i-1], layer_dims[i])) * 0.1
        parameters['b' + str(i)] = np.zeros((layer_dims[i], 1))
    return parameters
```

```python
initialize_parameters([4, 3, 2])
```

<h1>For Regression Problem</h1>

```python
def linear_forward(A, W, b):
    Z = np.dot(W.T, A) + b
    return Z
```

```python
def forward_propagation(X, parameters):
    A = X
    l = len(parameters) // 2  # number of layers in the network
    for i in range(1, l + 1):
        A_prev = A
        W = parameters['W' + str(i)]
        b = parameters['b' + str(i)]
        #print("A"+str(i-1)+":", A_prev)
        #print("W"+str(i)+":", W)
        #print("b"+str(i)+":", b)

        A = linear_forward(A_prev, W, b)
        #print("A"+str(i)+":", A)
        #print("*"*20)
    return A, A_prev

```

```python
X = df[['cgpa', 'profile_score']].values[0].reshape(2, 1)
Y = df[['lpa']].values[0][0]
```

```python
Y
```

```python
forward_propagation(X, initialize_parameters([2, 2, 1]))
```

```python
def update_parameters(parameters, y, y_hat, A, x):
    parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * 2 * (y-y_hat) * A[0][0])
    parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * 2 * (y-y_hat) * A[1][0])
    parameters['b2'][0][0] = parameters['b2'][0][0] + (0.001 * 2 * (y-y_hat))

    parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * 2 * (y-y_hat) * parameters['W2'][0][0] * x[0][0])
    parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * 2 * (y-y_hat) * parameters['W2'][0][0] * x[1][0])
    parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * 2 * (y-y_hat) * parameters['W2'][0][0])

    parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * 2 * (y-y_hat) * parameters['W2'][1][0] * x[0][0])
    parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * 2 * (y-y_hat) * parameters['W2'][1][0] * x[1][0])
    parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * 2 * (y-y_hat) * parameters['W2'][1][0])
```

```python
parameters = initialize_parameters([2, 2, 1])
epochs =5

for i in range(epochs):
    print("Epoch:", i+1)
    print("="*30)

    Loss =[]

    for j in range(len(df)):
        x = df[['cgpa', 'profile_score']].values[j].reshape(2, 1)
        y = df[['lpa']].values[j][0]

        y_hat, A = forward_propagation(x, parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameters, y, y_hat, A, x)
        loss = (y_hat - y) ** 2
        Loss.append(loss)

    print("Epoch - ", i+1, " Loss - ", np.array(Loss).mean())
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
def linear_forward(A, W, b):
    Z = np.dot(W.T, A) + b
    A = sigmoid(Z)
    return A
```

```python
def update_parameters(parameters, y, y_hat, A, x):
    parameters['W2'][0][0] = parameters['W2'][0][0] + (0.001 * (y-y_hat) * A[0][0])
    parameters['W2'][1][0] = parameters['W2'][1][0] + (0.001 * (y-y_hat) * A[1][0])
    parameters['b2'][0][0] = parameters['b2'][0][0] + (0.001 * (y-y_hat))

    parameters['W1'][0][0] = parameters['W1'][0][0] + (0.001 * (y-y_hat) * parameters['W2'][0][0] * A[0][0]*(1-A[0][0]) * x[0][0])
    parameters['W1'][0][1] = parameters['W1'][0][1] + (0.001 * (y-y_hat) * parameters['W2'][0][0] * A[0][0]*(1-A[0][0]) * x[1][0])
    parameters['b1'][0][0] = parameters['b1'][0][0] + (0.001 * (y-y_hat) * parameters['W2'][0][0])

    parameters['W1'][1][0] = parameters['W1'][1][0] + (0.001 * (y-y_hat) * parameters['W2'][1][0] * A[1][0]*(1-A[1][0]) * x[0][0])
    parameters['W1'][1][1] = parameters['W1'][1][1] + (0.001 * (y-y_hat) * parameters['W2'][1][0] * A[1][0]*(1-A[1][0]) * x[1][0])
    parameters['b1'][1][0] = parameters['b1'][1][0] + (0.001 * (y-y_hat) * parameters['W2'][1][0])
```

```python
parameters = initialize_parameters([2, 2, 1])
epochs = 50

for i in range(epochs):
    Loss=[]
    for j in range(len(df)):
        x = df[['cgpa', 'profile_score']].values[j].reshape(2, 1)
        y = df[['placed']].values[j][0]

        y_hat, A = forward_propagation(x, parameters)
        y_hat = y_hat[0][0]

        update_parameters(parameters, y, y_hat, A, x)
        loss = (-y_hat * np.log(y_hat) - (1-y_hat) * np.log(1-y_hat))
        Loss.append(loss)
    print("Epoch - ", i+1, " Loss - ", np.array(Loss).mean())
```
