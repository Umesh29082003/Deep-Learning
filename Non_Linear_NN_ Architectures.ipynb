```python
from keras.models import Model
```

```python
from keras.layers import *
```

<h1>Single Input Multiple Outputs</h1>

```python
x = Input(shape=(3,))
hidden1 = Dense(128, activation = 'relu')(x)        #At the end in the bracket = input from layer_name
hidden2 = Dense(64, activation = 'relu')(hidden1) 

output1 = Dense(1, activation = 'linear')(hidden2)
output2 = Dense(1, activation = 'sigmoid')(hidden2)
```

```python
model = Model(inputs = x , outputs = [output1, output2])
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model, show_layer_activations = True, show_layer_names= True, show_shapes= True)
```

<h1>Multiple Inputs and Single Output<h1>

```python
inputA = Input(shape = (32,))
inputB = Input(shape = (128,))

x = Dense(8, activation = 'relu')(inputA)
x1 = Dense(4, activation = 'relu')(x)

y = Dense(64, activation = 'relu')(inputB)
y1 = Dense(32, activation = 'relu')(y)
y2 = Dense(4, activation = 'relu')(y1)
```

```python
combined = concatenate([x1, y2])
z = Dense(2, activation = 'relu')(combined)
z1 = Dense(1, activation = 'linear')(z)
```

```python
model = Model(inputs = [inputA, inputB], outputs = z1)
```

```python
plot_model(model, show_shapes=True)
```
