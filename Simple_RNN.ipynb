#Integer Encoding

```python
import numpy as np

docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from keras.layers import TextVectorization
vectorizer = TextVectorization(output_mode="int")
```

```python
vectorizer.adapt(docs)
```

```python
vectorizer.get_vocabulary()
```

```python
vectorizer.vocabulary_size()
```

```python
sequences = vectorizer(docs)
print(sequences)
```

#Using Real Data

```python
from keras.datasets import imdb
from keras.preprocessing import sequence
from keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
x_train.shape, x_test.shape
```

```python
x_train
```

```python
len(x_train[0]), len(x_train[1]), len(x_train[2])
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
x_train = pad_sequences(x_train, maxlen=50)
x_test = pad_sequences(x_test, maxlen=50)
```

```python
x_train.shape
```

```python
model = Sequential()
model.add(SimpleRNN(32, input_shape=(50,1), return_sequences=False))
# If return_sequences = True then we send output at every return also
# If flase then only y hat comes out
model.add(Dense(1, activation='sigmoid'))
model.summary()

#Ip layer has 1 input although 50 timestamps
# Total params = 1*32 + 32*32(rec hidden) + 32 (bias) = 1088
# Op layers 32 + 1 bias = 33
```

```python
model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
model.fit(x_train, y_train, epochs = 5, validation_data = (x_test, y_test))
```
