| **32-bit Integer**        | `torch.int32`     | Standard signed integer type. Commonly used for indexing and general-purpose numerical tasks.                                                                                  |
| **64-bit Integer**        | `torch.int64`     | Long integer type. Often used for large indexing arrays or for tasks involving large numbers.                                                                                  |
| **8-bit Unsigned Integer**| `torch.uint8`     | 8-bit unsigned integer. Commonly used for image data (e.g., pixel values between 0 and 255).                                                                                    |

| **Complex 64**            | `torch.complex64` | Complex number type with 32-bit real and 32-bit imaginary parts. Used for scientific and signal processing tasks.                                                               |
| **Complex 128**           | `torch.complex128`| Complex number type with 64-bit real and 64-bit imaginary parts. Offers higher precision but uses more memory.                                                                 |

##4. Matrix Operation

```python
f = torch.randint(size = (2,3), low = 1, high = 5, dtype = torch.float32)
g = torch.randint(size = (3,2), low = 1, high = 5, dtype = torch.float32)

print(f)
print(g)
```

```python
# matric multiplication
print(torch.matmul(f,g))

print(torch.matmul(g,f))
```

```python
vector1 = torch.tensor([1,2])
vector2 = torch.tensor([3,4])

#dot product
torch.dot(vector1,vector2)
```

```python
#transpose - > basically swaps two dimensions
# we are swaping dimension 0 with one
print(f)
print(torch.transpose(f, 0, 1))

i = torch.randint(size = (2,3,4), low = 0, high = 10, dtype = torch.float32)
print(i)
print(torch.transpose(i, 0, 1))
```

```python
#determinant
print(torch.matmul(g,f))
print(torch.det(torch.matmul(g,f)))

#inverse
print(torch.inverse(torch.matmul(g,f)))
```

##5. Comparision Operation

```python
i = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
j = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)

print(i)
print(j)

# greater than
print(i>j)

# less than
print(i<j)

# equal to
print(i==j)

# not equal to
print(i!=j)

# greater than equal to
print(i>=j)

# less than equal to
print(i<=j)
```

# Inplace Operations

```python
m = torch.rand(2,3)
n = torch.rand(2,3)

print(m)
print(n)

```

```python
# if we write m + n then it creates a ew tensor of same size
# this is a problem when we do this for large datasets as matrix
m+n
```

```python
# to resolve this problem we sue inplace operations
m.add_(n)  # Now the sum is stored in m itself
print(m)
```

```python
# Sigmoid example
m.sigmoid_()
m
```

#CPU vs GPU

```python
import torch
import time

size = 10000

matrix_cpu1 = torch.rand(size, size)
matrix_cpu2 = torch.rand(size, size)

matrix_gpu1 = matrix_cpu1.to('cuda')
matrix_gpu2 = matrix_cpu2.to('cuda')

#Measure Time on CPU
start_time = time.time()
result_cpu = torch.matmul(matrix_cpu1, matrix_cpu2)
end_time = time.time()
cpu_time = end_time - start_time

#Measure Time on GPU
start_time = time.time()
result_gpu = torch.matmul(matrix_gpu1, matrix_gpu2)
torch.cuda.synchronize() #To ensure that all GPU operations are complete
end_time = time.time()
gpu_time = end_time - start_time

print(f'CPU Time: {cpu_time:.4f} seconds')
print(f'GPU Time: {gpu_time:.4f} seconds')

print("\nSpeed Up (CPU time/ GPU time):", cpu_time/gpu_time)
```
